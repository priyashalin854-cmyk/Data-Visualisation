```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("shopify_stock.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
df.isnull().sum()
```

```python
df.dropna()
```

```python
plt.plot(df['date'],df['close'])
plt.title("Shopify's Close price over time")
plt.xlabel("Close Price")
plt.ylabel("Time")
plt.show()
```

```python
df['daily_delta']=df['close'] - df['open']
```

```python
df['daily_return']=(df['close']-df['open'])/df['open']
```

```python
df.head()
```

```python
df.tail()
```

```python
plt.hist(df['daily_return'])
plt.title("Shopify's Daily Return Histogram")
plt.xlabel("Daily Return")
plt.ylabel("Frequency")
plt.show()
```
