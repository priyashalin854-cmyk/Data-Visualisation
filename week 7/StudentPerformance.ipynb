```python
df['Total Score']=df['math score']+df['reading score']+df['writing score']
```

```python
df['Percentage'] = (df['Total Score']/300)*100
```

```python
df[['Total Score','Percentage']].head()
```

```python
outlier_results = {}

for col in score_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
```

```python
outliers = df[
    (df[col] < lower_limit) |
    (df[col] > upper_limit)
]
outlier_results[col] = outliers
print("Subject:", col)
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Number of Outliers:", len(outliers))
```

```python
for subject, outliers in outlier_results.items():

    print("\nOutliers in", subject)

    if len(outliers) > 0:
        print(
            outliers[
                [
                    "gender",
                    "race/ethnicity",
                    "math score",
                    "reading score",
                    "writing score",
                    "Total Score",
                    "Percentage"
                ]
            ]
        )
    else:
        print("No outliers detected.")
```
